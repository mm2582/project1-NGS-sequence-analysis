## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """
    Validate a sequencing read against a reference sequence.

    Checks that the read is the same length as the reference and contains
    only valid DNA characters (A, C, G, T). Lowercase letters are converted
    to uppercase. Returns a dictionary with the uppercased sequence, a
    validity flag, and a list of any errors found.

    Args:
        sequence (str): The read sequence to validate (any length/content).
        reference (str): The reference sequence (uppercase A/C/G/T only).

    Returns:
        dict: Keys are 'sequence' (str, uppercased), 'valid' (bool),
              and 'errors' (list of str).
    """
    # Convert sequence to uppercase (repair lowercase reads)
    sequence = sequence.upper()

    # Start with an empty list of errors
    errors = []

    # Checking if the length matches the reference
    if len(sequence) != len(reference):
        errors.append("wrong_length")

    # Checking if all characters are valid DNA bases (A, C, G, T)
    valid_bases = set("ACGT")
    if not all(base in valid_bases for base in sequence):
        errors.append("invalid_characters")

    # Read is valid only if there are no errors
    is_valid = len(errors) == 0

    # Return the result as a dictionary
    return {
        "sequence": sequence,
        "valid": is_valid,
        "errors": errors
    }
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """
    Find all mismatches between a sequencing read and the reference sequence.

    Compares the read and reference base by base, returning a list of
    dictionaries describing each mismatch. Positions are numbered from 1.
    Returns an empty list if the read perfectly matches the reference.

    Args:
        sequence (str): The read sequence (valid, uppercase A/C/G/T, same length as reference).
        reference (str): The reference sequence (valid, uppercase A/C/G/T).

    Returns:
        list: A list of dictionaries, each with keys 'position' (int, 1-indexed),
              'reference_base' (str), and 'read_base' (str). Empty list if no mismatches.
    """
    # Start with an empty list to store mutations
    mutations = []

    # Loop through each position in the sequence
    for i in range(len(sequence)):
        # Compare the base in the read to the base in the reference
        if sequence[i] != reference[i]:
            # Found a mismatch — record it
            mutation = {
                "position": i + 1,              # +1 because positions start at 1, not 0
                "reference_base": reference[i], # What the reference has here
                "read_base": sequence[i]        # What the read has here
            }
            mutations.append(mutation)

    return mutations
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """Analyse all reads by validating and finding mutations.

    Uses validate_read and find_mutations to process each read.
    Returns a new list of dicts (does not modify the original reads).

    Args:
        reads: List of dicts with keys 'read_id', 'sample', 'group', 'sequence'.
        reference: A valid uppercase nucleotide string.

    Returns:
        A list of dicts, one per read, with keys:
            'read_id', 'sample', 'group', 'sequence' (uppercase),
            'status' ('valid' or 'invalid'), 'errors' (list),
            'mutations' (list of mismatch dicts, or [] if invalid).
        Returns [] if reads is empty.
    """
    if not reads:
      return []
    analysed = []

    for read in reads:
        # Validate the read
        validation = validate_read(read["sequence"], reference)

        # Build the output dictionary
        result = {
            "read_id": read["read_id"],
            "sample": read["sample"],
            "group": read["group"],
            "sequence": validation["sequence"],  # uppercase version
            "status": "valid" if validation["valid"] else "invalid",
            "errors": validation["errors"],
            "mutations": [],  # default to empty
        }

        # Only find mutations if the read is valid
        if validation["valid"]:
            result["mutations"] = find_mutations(validation["sequence"], reference)

        analysed.append(result)

    return analysed
````

````python
def summarise_samples(analysed_reads):
    """
    Summarise mismatch patterns between samples from analysed read data.

    Groups reads by sample, calculates statistics (valid/invalid counts,
    mismatch rates, substitution types, positions), and prints
    a summary table for each experimental group.

    Args:
        analysed_reads (list): Output from analyse_reads - list of dictionaries
                              with keys 'read_id', 'sample', 'group', 'sequence',
                              'status', 'errors', 'mutations'.

    Returns:
        dict: Nested dictionary with one entry per sample containing:
              'group', 'valid_reads', 'invalid_reads', 'mismatch_reads',
              'mismatch_percentage', 'mismatch_events', 'substitution_counts',
              'position_counts', 'top_positions'.
    """
    # empty input
    if not analysed_reads:
        return {}

    # Group reads by sample
    samples = {}
    for read in analysed_reads:
        sample_name = read['sample']
        if sample_name not in samples:
            samples[sample_name] = []
        samples[sample_name].append(read)

    # Calculate statistics for each sample
    summary = {}
    for sample_name, sample_reads in samples.items():
        summary[sample_name] = calculate_sample_stats(sample_reads)

    # Group samples by experimental group and print summaries
    groups = {}
    for sample_name, stats in summary.items():
        group_name = stats['group']
        if group_name not in groups:
            groups[group_name] = {}
        groups[group_name][sample_name] = stats

    # Print summary for each group
    for group_name in sorted(groups.keys()):
        print_group_summary(group_name, groups[group_name])

    return summary
````

````python
def calculate_sample_stats(sample_reads):
    """
    Calculate all statistics for a single sample's reads.

    Args:
        sample_reads (list): List of analysed read dictionaries for one sample.

    Returns:
        dict: Dictionary containing all calculated statistics for this sample.
    """
    # Count valid and invalid reads
    valid_reads = 0
    invalid_reads = 0
    mismatch_reads = 0  # Valid reads with at least one mutation
    mismatch_events = 0  # Total individual mutations
    substitution_counts = {}  # e.g., {"A>G": 9}
    position_counts = {}  # e.g., {10: 7, 25: 4}

    for read in sample_reads:
        if read['status'] == 'valid':
            valid_reads += 1

            # Check if this read has any mutations
            if len(read['mutations']) > 0:
                mismatch_reads += 1

            # Count each mutation
            for mutation in read['mutations']:
                mismatch_events += 1

                # Track substitution type (e.g., "A>G")
                ref_base = mutation['reference_base']
                read_base = mutation['read_base']
                substitution = f"{ref_base}>{read_base}"

                if substitution not in substitution_counts:
                    substitution_counts[substitution] = 0
                substitution_counts[substitution] += 1

                # Track position
                pos = mutation['position']
                if pos not in position_counts:
                    position_counts[pos] = 0
                position_counts[pos] += 1
        else:
            invalid_reads += 1

    # Calculate mismatch percentage (only for valid reads)
    if valid_reads > 0:
        mismatch_percentage = (mismatch_reads / valid_reads) * 100
    else:
        mismatch_percentage = None

    # Find top 3 positions with most mismatches
    # Sort by count (descending), then by position (ascending) for ties
    sorted_positions = sorted(position_counts.items(), key=lambda x: (-x[1], x[0]))
    top_positions = [pos for pos, count in sorted_positions[:3]]

    # Get the group name from the first read
    group = sample_reads[0]['group']

    return {
        "group": group,
        "valid_reads": valid_reads,
        "invalid_reads": invalid_reads,
        "mismatch_reads": mismatch_reads,
        "mismatch_percentage": mismatch_percentage,
        "mismatch_events": mismatch_events,
        "substitution_counts": substitution_counts,
        "position_counts": position_counts,
        "top_positions": top_positions
    }
````

````python
def print_group_summary(group_name, sample_stats):
    """
    Print a formatted summary table for all samples in one experimental group.

    Displays valid/invalid counts, mismatch percentage, total events, and
    top 3 mutation positions. Also prints a substitution mutations table
    showing counts and percentages per 100 valid reads for each substitution type.

    Args:
        group_name (str): Name of the experimental group (e.g., "treatment", "control").
        sample_stats (dict): Dictionary of {sample_name: stats_dict} for this group.
    """
    print(f"\nGroup: {group_name.upper()}")

    # Print main summary table header
    print(f"{'Sample':<12} {'Valid':<8} {'Invalid':<10} {'Mismatch%':<12} {'Events':<10} {'Top Positions'}")
    print(f"{'-'*12} {'-'*8} {'-'*10} {'-'*12} {'-'*10} {'-'*20}")

    # Print each sample's stats
    for sample_name in sorted(sample_stats.keys()):
        stats = sample_stats[sample_name]

        # Format mismatch percentage
        if stats['mismatch_percentage'] is not None:
            pct = f"{stats['mismatch_percentage']:.1f}%"
        else:
            pct = "N/A"

        # Format top positions
        top_pos = ", ".join(str(p) for p in stats['top_positions'])

        print(f"{sample_name:<12} {stats['valid_reads']:<8} {stats['invalid_reads']:<10} {pct:<12} {stats['mismatch_events']:<10} {top_pos}")

    # Print substitution mutations table
    all_substitutions = set()
    for stats in sample_stats.values():
        all_substitutions.update(stats['substitution_counts'].keys())

    if all_substitutions:
        print(f"\nSubstitution Mutations (count and per 100 valid reads):")

        # Get sorted sample names for consistent column order
        sample_names = sorted(sample_stats.keys())

        # Print header row
        header = f"{'Substitution':<15}"
        for name in sample_names:
            header += f" {name:>12}"
        print(header)
        print(f"{'-'*15} {'-'*12*len(sample_names)}")

        # Print one row per substitution type (sorted alphabetically)
        for sub in sorted(all_substitutions):
            row = f"{sub:<15}"
            for name in sample_names:
                stats = sample_stats[name]
                count = stats['substitution_counts'].get(sub, 0)
                valid = stats['valid_reads']
                if valid > 0:
                    pct = (count / valid) * 100
                    row += f" {count:>6} ({pct:>4.1f}%)"
                else:
                    row += f" {count:>6} (N/A)"
            print(row)

    print()
````

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:


Each list must contain **one percentage per biological sample**.

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

````python
def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
    """
    Compare mismatch frequencies at top positions between control and treated groups.

    Combines position counts across samples to find the top 3 positions,
    calculates per-sample mismatch percentages at each position, and performs
    a Welch's t-test to compare groups.

    Args:
        sample_summary (dict): Output from summarise_samples function.
        control_group (str): Name of the control group (default: "control").
        treated_group (str): Name of the treated group (default: "treatment").

    Returns:
        list: One dictionary per top position with keys 'position', 'control_n',
              'treated_n', 'control_mean', 'treated_mean', 'difference',
              'statistic', 'p_value'.
    """
    # Separate samples by group
    control_samples = {}
    treated_samples = {}

    for sample_name, stats in sample_summary.items():
        if stats['group'] == control_group:
            control_samples[sample_name] = stats
        elif stats['group'] == treated_group:
            treated_samples[sample_name] = stats

    # Combine position counts across all samples to find top positions
    combined_counts = {}

    for sample_name, stats in control_samples.items():
        for pos, count in stats['position_counts'].items():
            if pos not in combined_counts:
                combined_counts[pos] = 0
            combined_counts[pos] += count

    for sample_name, stats in treated_samples.items():
        for pos, count in stats['position_counts'].items():
            if pos not in combined_counts:
                combined_counts[pos] = 0
            combined_counts[pos] += count

    # Find top 5 positions
    sorted_positions = sorted(combined_counts.items(), key=lambda x: (-x[1], x[0]))
    top_positions = [pos for pos, count in sorted_positions[:5]]

    # For each top position, calculate per-sample percentages and run stats
    results = []

    for position in top_positions:
        result = analyse_position(
            position, control_samples, treated_samples
        )
        results.append(result)

    # Print results in scientific style
    print_results(results)

    return results
````

````python
def analyse_position(position, control_samples, treated_samples):
    """
    Calculate mismatch percentages for one position across all samples,
    then compare groups statistically.

    Args:
        position (int): The position to analyse.
        control_samples (dict): {sample_name: stats} for control group.
        treated_samples (dict): {sample_name: stats} for treated group.

    Returns:
        dict: Statistics for this position.
    """
    # Calculate mismatch percentage for each control sample
    control_percentages = []
    control_n = 0
    for sample_name, stats in control_samples.items():
        if stats['valid_reads'] > 0:
            count = stats['position_counts'].get(position, 0)
            pct = (count / stats['valid_reads']) * 100
            control_percentages.append(pct)
            control_n += 1

    # Calculate mismatch percentage for each treated sample
    treated_percentages = []
    treated_n = 0
    for sample_name, stats in treated_samples.items():
        if stats['valid_reads'] > 0:
            count = stats['position_counts'].get(position, 0)
            pct = (count / stats['valid_reads']) * 100
            treated_percentages.append(pct)
            treated_n += 1

    # Calculate means
    control_mean = sum(control_percentages) / len(control_percentages) if control_percentages else None
    treated_mean = sum(treated_percentages) / len(treated_percentages) if treated_percentages else None

    # Calculate difference (treated - control)
    if control_mean is not None and treated_mean is not None:
        difference = treated_mean - control_mean
    else:
        difference = None

    # Run statistical test
    if control_percentages and treated_percentages:
        stat_result = calculate_statistics(control_percentages, treated_percentages)
    else:
        stat_result = {"statistic": None, "p_value": None}

    return {
        "position": position,
        "control_n": control_n,
        "treated_n": treated_n,
        "control_mean": control_mean,
        "treated_mean": treated_mean,
        "difference": difference,
        "statistic": stat_result["statistic"],
        "p_value": stat_result["p_value"]
    }
````

````python
def print_results(results):
    """
    Print statistical comparison results in scientific style.

    Args:
        results (list): List of position result dictionaries from analyse_position.
    """
    print("\nStatistical Comparison of Top Mismatch Positions")
    print("(Two-sided Welch's t-test)\n")

    for r in results:
        pos = r['position']
        ctrl_mean = r['control_mean']
        treat_mean = r['treated_mean']
        diff = r['difference']
        ctrl_n = r['control_n']
        treat_n = r['treated_n']
        p_val = r['p_value']

        # Format p-value
        if p_val is not None:
            if p_val < 0.001:
                p_str = "p < 0.001"
            else:
                p_str = f"p = {p_val:.3f}"
            stat_str = f"t = {r['statistic']:.2f}"
        else:
            p_str = "p = N/A"
            stat_str = "statistic = N/A"

        # Format means
        if ctrl_mean is not None:
            ctrl_str = f"{ctrl_mean:.2f}%"
        else:
            ctrl_str = "N/A"
        if treat_mean is not None:
            treat_str = f"{treat_mean:.2f}%"
        else:
            treat_str = "N/A"
        if diff is not None:
            diff_str = f"{diff:.2f}"
        else:
            diff_str = "N/A"

        print(f"Position {pos}: Control mean = {ctrl_str} (n={ctrl_n}), "
              f"Treated mean = {treat_str} (n={treat_n}), "
              f"Difference = {diff_str} percentage points, "
              f"{stat_str}, {p_str}")

    print()
````
